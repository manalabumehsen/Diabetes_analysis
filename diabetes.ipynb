## Data Science
</div>

Mit Colab können Sie das volle Potenzial der beliebten Python-Bibliotheken nutzen, um Daten zu analysieren und visualisieren. Die Codezelle unten verwendet <strong>NumPy</strong>, um zufällige Daten zu erstellen, und <strong>Matplotlib</strong>, um sie zu visualisieren. Klicken Sie zum Bearbeiten des Codes einfach auf die Zelle und beginnen Sie mit der Bearbeitung.

Sie können Ihre eigenen Daten aus Ihrem Google Drive-Konto in Colab-Notebooks importieren, beispielsweise aus Tabellen sowie aus Github und vielen anderen Quellen. Weitere Informationen zum Importieren von Daten und dazu, wie Colab für Data Science genutzt werden kann, finden Sie unten in den Links unter <a href="#working-with-data">Mit Daten arbeiten</a>.

```
import numpy as np
import IPython.display as display
from matplotlib import pyplot as plt
import io
import base64

ys = 200 + np.random.randn(100)
x = [x for x in range(len(ys))]

fig = plt.figure(figsize=(4, 3), facecolor='w')
plt.plot(x, ys, '-')
plt.fill_between(x, ys, 195, where=(ys > 195), facecolor='g', alpha=0.6)
plt.title("Sample Visualization", fontsize=10)

data = io.BytesIO()
plt.savefig(data)
image = F"data:image/png;base64,{base64.b64encode(data.getvalue()).decode()}"
alt = "Sample Visualization"
display.display(display.Markdown(F"""![{alt}]({image})"""))
plt.close(fig)
```

```
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
df.info()
(df.describe())
print(df.isnull().sum())
duplictes = df.duplicated().sum()
print(duplictes)
if 'Diabetes_012' in df.columns :
 sns.countplot(x='Diabetes_012',data =df)
 plt.show()
```

```
df.describe()
numeric_cols=df.select_dtypes(include= ['int64','float64']).columns
for col in numeric_cols:
  plt.figure(figsize=(6,4))
  sns.histplot(df[col],kde=True, bins=30)
  plt.title(f'distribution of {col}')
  plt.show()

  categorical_cols=df.select_dtypes(include=['object']).columns
  for col in categorical_cols :
    plt.figure(figsize=(6,4))
    sns.countplot(x=col,data=df)
    plt.title(f'count of {col}')
    plt.show()
```

```
for col in numeric_cols:
  plt.figure(figsize=(12,8))
  sns.boxplot(x='Diabetes_012',y=col,data=df)
  plt.title(f'{col} vs diabetes_012')
  plt.show()
  plt.figure(figsize=(12,8))
  sns.heatmap(df[numeric_cols].corr(),annot=True,cmap='coolwarm')
plt.title('correltion heatmap')
plt.show()
```

```
mean_per_group =df.groupby('Diabetes_012').agg({col:'mean' for col in numeric_cols})
display(mean_per_group)
```
